### Import Relevant Libraries

### Merge the Datasets into Master Dataset

```python
# Store the datasets into variables
df_fake = pd.read_csv("../../Datasets/Dataset1_fake_counts_cleaned_1.csv")
df_true = pd.read_csv("../../Datasets/Dataset1_true_counts_cleaned_1.csv")
```

```python
df_all = pd.concat([df_fake, df_true], ignore_index=True)
df_all
```

### Shuffle the Order of the Dataset

```python
df_all = df_all.sample(frac=1, random_state=42).reset_index(drop=True)
df_all
```
