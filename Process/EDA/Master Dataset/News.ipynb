## Exploratory Data Analysis of Master Dataset - Datasets 1 and 2

### Import the relevant libraries

```python
# For dataset manipulation
import pandas as pd
```

### Merge the datasets into a master dataset

```python
# Store the datasets into variables
df_fake = pd.read_csv("../../../Datasets/Dataset1_fake_counts.csv")
df_true = pd.read_csv("../../../Datasets/Dataset1_true_counts.csv")
df_mixed = pd.read_csv("../../../Datasets/Dataset2_mixed_counts.csv")
```

```python
df_all = pd.concat([df_fake, df_true, df_mixed], ignore_index=True)
```

```python
# Show the master dataset
df_all
```

### Save the master dataset

```python
df_all.to_csv('../../../Datasets/Dataset_master_counts.csv', index=False)
```

### Combining Title + Text

```python
df_all = df_all.drop(columns=[
    'title_word_count', 
    'title_punct_count',
    'title_readability_ease',
    'text_word_count', 
    'text_punct_count',
    'text_readability_ease'
]).reset_index(drop=True)
```

```python
df_all['combined_data'] = df_all['title'] + ' ' + df_all['text']
```

```python
df_all = df_all[["title", "text", "combined_data", "label"]]
```

```python
df_all
```

### Save the Updated Dataset

```python
df_all.to_csv('../../../Datasets/Dataset_master_combineddata.csv', index=False)
```
