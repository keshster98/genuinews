## Balanced Fake/True, Combined Text + Title Param Counts
___


### Import the Relevant Libraries

```python
# For dataset manipulation
import numpy as np
import pandas as pd
# For punctuation count
import spacy
# For readability ease count
import textstat
# For progress bar
from tqdm.notebook import tqdm
tqdm.pandas()
```

### Load The spaCy English Model

```python
nlp = spacy.load("en_core_web_sm")
```

### Custom Functions

```python
# Removes whitespaces and lowers case
def whitespace_lower(component):
    return component.str.lower().str.split().str.join(' ')

# Calculate total words
def calc_word(component):
    return component.astype(str).apply(lambda x: len(x.split()))

# Calculate total punctuations
def calc_punct(component):
    doc = nlp(component)
    # For every punct found, sum increases by 1
    return sum(1 for token in doc if token.is_punct) 

# Calculate readability ease
def calc_readability(component):
    return textstat.flesch_reading_ease(component)
```

### Store the Datasets into Variables

```python
df_d1_fake = pd.read_csv("../../Datasets/Dataset1_fake_counts.csv")
df_d1_true = pd.read_csv("../../Datasets/Dataset1_true_counts.csv")
df_d2_mixed = pd.read_csv("../../Datasets/Dataset2_mixed_counts.csv")
```

### Reconvert Any "" That Was Previously NaN in Dataset 2

```python
df_d2_mixed['title'] = df_d2_mixed['title'].replace("", np.nan)
df_d2_mixed['text'] = df_d2_mixed['text'].replace("", np.nan)
```

### Drop NaN (Null) Values

```python
df_d2_mixed.isna().sum()
```

```python
df_d2_mixed = df_d2_mixed.dropna(subset=['title', 'text']).reset_index(drop=True)
```

```python
df_d2_mixed.isna().sum()
```

### Separate Dataset 2 into Fake and True

```python
# Retrieve only fake news from Dataset 2
df_d2_fake = df_d2_mixed[df_d2_mixed['label'] == 0].reset_index(drop=True)

# Retrieve only true news from Dataset 2
df_d2_true = df_d2_mixed[df_d2_mixed['label'] == 1].reset_index(drop=True)
```

### Lower Case and Remove Whitespaces

```python
# Dataset 1 - Fake
df_d1_fake['title'] = whitespace_lower(df_d1_fake['title'])
df_d1_fake['text'] = whitespace_lower(df_d1_fake['text'])

# Dataset 1 - True
df_d1_true['title'] = whitespace_lower(df_d1_true['title'])
df_d1_true['text'] = whitespace_lower(df_d1_true['text'])

# Dataset 2 - Fake 
df_d2_fake['title'] = whitespace_lower(df_d2_fake['title'])
df_d2_fake['text'] = whitespace_lower(df_d2_fake['text'])

# Dataset 2 - True
df_d2_true['title'] = whitespace_lower(df_d2_true['title'])
df_d2_true['text'] = whitespace_lower(df_d2_true['text'])
```

### Dropping Duplicates for Dataset 1 - Fake

```python
df_d1_fake.duplicated(subset=['title', 'text'], keep="first").sum() # keep="first" ensures the first occurence of the duplicate is kept
```

```python
# Dropping rows where both title and text are exact duplicates after the first occurence
df_d1_fake = df_d1_fake.drop_duplicates(subset=['title', 'text'], keep='first').reset_index(drop=True)
```

### Dropping Duplicates for Dataset 1 - True

```python
df_d1_true.duplicated(subset=['title', 'text'], keep="first").sum()
```

```python
df_d1_true = df_d1_true.drop_duplicates(subset=['title', 'text'], keep='first').reset_index(drop=True)
```

### Dropping Duplicates for Dataset 2 - Fake

```python
df_d2_fake.duplicated(subset=['title', 'text'], keep="first").sum()
```

```python
df_d2_fake = df_d2_fake.drop_duplicates(subset=['title', 'text'], keep='first').reset_index(drop=True)
```

### Dropping Duplicates for Dataset 2 - True

```python
df_d2_true.duplicated(subset=['title', 'text'], keep="first").sum()
```

```python
df_d2_true = df_d2_true.drop_duplicates(subset=['title', 'text'], keep='first').reset_index(drop=True)
```

### Merging the Dataset

```python
df_all = pd.concat([df_d1_fake, df_d1_true, df_d2_fake, df_d2_true], ignore_index=True)
```

```python
df_all
```

```python
df_all["label"].value_counts()
```

### Shuffle the Order of the Dataset

```python
df_all = df_all.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
df_all
```

### Save the Updated Dataset (Backup)

```python
df_all.to_csv('../../Datasets/Dataset_master_counts_cleanrows.csv', index=False)
```
